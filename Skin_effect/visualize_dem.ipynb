```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

path = "/Users/dorageeraerts/Documents/PhD/simulation/Muraves/Simulation_Studies/PUMAS-and-TURTLE/Mulder/vesuvio_5m_cut.asc"
# Read ESRI ASCII grid
with open(path) as f:
    header = [next(f) for _ in range(6)]

data = np.loadtxt(path, skiprows=6)

x = np.arange(data.shape[1])
y = np.arange(data.shape[0])
X, Y = np.meshgrid(x, y)

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(
    X, Y, data,
    cmap="terrain",
    linewidth=0
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Elevation")

plt.show()
```
